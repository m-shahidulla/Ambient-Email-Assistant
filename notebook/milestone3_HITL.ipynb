```python
import pandas as pd
df = pd.read_csv("../data/sample_emails_with_triage_200.csv")
df.head()
```

```python


def email_assistant(body):
    text = body.lower()

    if "urgent" in text or "submit" in text or "deadline" in text:
        return {"action" : "notify" , "tone" : "urgent"}
    elif "thank you" in text or "thanks" in text:
        return {"action" : "ignore" , "tone": "polite"}
    else:
        return {"action" :"respond",  "tone"  :"neutral"}
```

```python
DANGEROUS_ACTIONS = ["respond"]

def hitl_check(action):
    if action in DANGEROUS_ACTIONS:
        return "WAIT_FOR_HUMAN" 
    return "AUTO_APPROVED"

#Simulate the human approval
def human_decision():
    decision = input("Approve action? (yes/no): ")
    return decision.lower() == "yes"
```

```python
import pandas as pd

results = []
for _, row in df.sample(5).iterrows():
    action, tone = email_assistant(row["body"])

    status = hitl_check(action)
    if status == "WAIT_FOR_HUMAN":
        approved = human_decision()
        final_action = action if approved else "blocked"
    else:
        final_action = action
    
    results.append({
        "email" : row["body"],
        "ai_action" : action,
        "final_action": final_action,
        "hitl_status" : status
    })
pd.DataFrame(results)
results_df = pd.DataFrame(results)
```
